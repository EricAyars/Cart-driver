# Single-cart analysis
Data from 20260519 and 20260520. Sweep through frequencies and look at amplitude and phase of cart's motion.

Cart is held between two PASCO springs, resonance frequency about 0.75Hz (rough measurement with stopwatch.) Sweep from 0.2 Hz -> 1.2 Hz by 0.01 Hz.

Files:
* 20260519A.txt Servo drive, amplitude = 2.0 (cm, roughly). Cart disconnected from springs at peak, data probably not good.
* 20260519B.txt Speaker drive, amplitude = 10V (0.7-0.8mm). Cart battery died at 1.15Hz, so incomplete scan but usable. Phase data in the file, but it's not meaningful since no drive phase available for cocmparison.
* 20260519C.txt (mislabeled, should be 20260520A.txt) Servo drive, amplitude = 0.45. Started 0830 5/20/26.


```python
from pylab import *
```

```python
frequencyA, ampA, phaseA = loadtxt("20260519A.txt", skiprows=7, unpack=True)
frequencyB, ampB, phaseB = loadtxt("20260519B.txt", skiprows=7, unpack=True)
```

```python
semilogy(frequencyA, ampA, 'r.')
title("Servo sweep @ amp=2.0")
xlabel("Frequency (Hz)")
ylabel("Cart Oscillation Amplitude (cm)")
```

```python
plot(frequencyA, ampA, 'r.')
title("Servo sweep @ amp=2.0")
xlabel("Frequency (Hz)")
ylabel("Cart Oscillation Amplitude (cm)")
```

```python
semilogy(frequencyB, ampB, 'r.')
title("Speaker sweep @ amp=10V")
xlabel("Frequency (Hz)")
ylabel("Cart Oscillation Amplitude (cm)")
```

```python
plot(frequencyB, ampB, 'r.')
title("Speaker sweep @ amp=10V")
xlabel("Frequency (Hz)")
ylabel("Cart Oscillation Amplitude (cm)")
```

```python
fig, ax1 = subplots()
ax1.set_title("Speaker sweep @ amp=10V\nAmplitude and Phase")

ax1.semilogy(frequencyB, ampB, 'r-')
ax1.set_xlabel("Frequency (Hz)")
ax1.set_ylabel("Cart Oscillation Amplitude (cm)")

ax2 = ax1.twinx()
ax2.plot(frequencyB, phaseB, 'g.')
ax2.set_ylabel("Cart phase")
```

```python
fig, ax1 = subplots()
ax1.set_title("Servo sweep @ amp=2.0\nAmplitude and Phase")

ax1.semilogy(frequencyA, ampA, 'r-')
ax1.set_xlabel("Frequency (Hz)")
ax1.set_ylabel("Cart Oscillation Amplitude (cm)")

ax2 = ax1.twinx()
ax2.plot(frequencyA, phaseA, 'g.')
ax2.set_ylabel("Cart phase")
```

```python
fig, ax1 = subplots()
ax1.set_title("Speaker sweep @ amp=10V\nAmplitude and Phase")

ax1.plot(frequencyB, ampB, 'r-')
ax1.set_xlabel("Frequency (Hz)")
ax1.set_ylabel("Cart Oscillation Amplitude (cm)")

ax2 = ax1.twinx()
ax2.plot(frequencyB, phaseB, 'g.')
ax2.set_ylabel("Cart phase")
```

Note that the phase information in the previous graph is bogus since to phase information is actually recorded for the speaker driver.

```python
frequencyC, ampC, phaseC = loadtxt("20260520C.txt", skiprows=7, unpack=True)
```

```python
fig, ax1 = subplots()
ax1.set_title("Servo sweep @ amp=0.45cm\nAmplitude and Phase")

ax1.plot(frequencyC, ampC, 'r-')
ax1.set_xlabel("Frequency (Hz)")
ax1.set_ylabel("Cart Oscillation Amplitude (cm)")

ax2 = ax1.twinx()
ax2.plot(frequencyC, phaseC, 'g.')
ax2.set_ylabel("Cart phase")
```

```python
for j in range(len(phaseC)):
    if phaseC[j] < 0:
        phaseC[j] += 2*pi
    if phaseC[j] > 2*pi:
        phaseC[j] -= 2*pi
        
```

```python
fig, ax1 = subplots()
ax1.set_title("Servo sweep @ amp=0.45cm\nAmplitude and Phase")

ax1.plot(frequencyC, ampC, 'r-')
ax1.set_xlabel("Frequency (Hz)")
ax1.set_ylabel("Cart Oscillation Amplitude (cm)")

ax2 = ax1.twinx()
ax2.plot(frequencyC, phaseC, 'g.')
ax2.set_ylabel("Cart phase")
```

```python
fig, ax1 = subplots()
ax1.set_title("Servo sweep @ amp=0.45cm\nAmplitude and Phase")

ax1.semilogy(frequencyC, ampC, 'r-')
ax1.set_xlabel("Frequency (Hz)")
ax1.set_ylabel("Cart Oscillation Amplitude (cm)")

ax2 = ax1.twinx()
ax2.plot(frequencyC, phaseC, 'g.')
ax2.set_ylabel("Cart phase")
```
